```python
from duckduckgo_search import DDGS
from transformers import pipeline
import requests
from bs4 import BeautifulSoup
import time

# Load the summarizer
summarizer = pipeline("summarization")

def extract_text_from_url(url):
    try:
        response = requests.get(url, timeout=10)
        soup = BeautifulSoup(response.text, "html.parser")
        paragraphs = soup.find_all("p")
        text = " ".join(p.get_text() for p in paragraphs)
        return text.strip()
    except Exception as e:
        print(f"Error fetching article: {e}")
        return None

def summarize_topic(topic, max_results=3):
    print(f"\n🔎 Researching topic: {topic}\n")
    with DDGS() as ddgs:
        results = ddgs.text(topic, max_results=max_results)

        for i, r in enumerate(results, 1):
            title = r['title']
            url = r['href']
            print(f"--- Article {i} ---")
            print(f"Title: {title}")
            print(f"Link: {url}")

            article_text = extract_text_from_url(url)
            if not article_text or len(article_text) < 200:
                print("⚠️ Skipping article (too short or couldn't extract)")
                continue

            # Summarize only first 1,000 tokens to avoid model limits
            text_to_summarize = article_text[:2000]

            try:
                summary = summarizer(text_to_summarize, max_length=100, min_length=40, do_sample=False)
                print("✅ Summary:")
                print(summary[0]['summary_text'])
            except Exception as e:
                print("❌ Could not summarize:", e)

            print("\n")  # Add space between articles
            time.sleep(2)  # Pause between requests to be polite

# Try it with a topic of your choice!
summarize_topic("AI in education")
```
